```python
%load_ext autoreload
%autoreload 2
from prl.core import Dataset
```

```python
dataset = Dataset("roi_train3")
```

```python
dataset.create_datalist(rebuild=True)
```

```python
for split, g in dataset.cases.groupby("split"):
    print(split)
    print("\tPRL: ", g.case_type.eq("PRL").sum())
    print("\tLesions: ", g.case_type.eq("Lesion").sum())

    print("\tSubject ID's", g.index.get_level_values("subid").unique().values)
```

```python
import pandas as pd
dataset.cases['isPRL'] = dataset.cases.case_type.eq("PRL")
dataset.cases['isLesion'] = dataset.cases.case_type.eq("Lesion")

table = (
    dataset.cases.groupby(level="subid")
    .agg(
        nPRL=pd.NamedAgg("isPRL", aggfunc=sum),
        nLesion=pd.NamedAgg("isLesion", aggfunc=sum)
    )
)

print(table.to_markdown())
```
